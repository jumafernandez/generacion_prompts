```python
import pandasai as pai
from pandasai_openai import OpenAI

llm = OpenAI(api_token="")

# Set your OpenAI API key
pai.config.set({"llm": llm})
```

```python
# Load your data
df = pai.read_csv("https://raw.githubusercontent.com/jumafernandez/generacion_prompts/refs/heads/main/semana6/dataset_calidad_de_vida.csv")

df
```

```python
response1 = df.chat("¿Cuales son los paises con mejor calidad de vida?")
```

```python
print(response1)
```

```python
response2 = df.chat("¿Podes hacer un grafico de barras con los paises con mejor calidad de vida?")
```

```python
response3 = df.chat("¿Podes hacer un grafico de barras con los paises con mejor calidad de vida, cada barra debe tener un color distinto, el titulo debe ser indice de calidad de vida?")
```
